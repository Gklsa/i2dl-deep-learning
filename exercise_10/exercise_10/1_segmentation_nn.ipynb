Semantic Segmentation 
============

In this task, you will be focusing on semantic segmentation, a computer vision technique. Unlike image classification, the objective here is not to classify the entire image as a whole, but rather to classify each individual pixel within the image. Consequently, the network's output is not a single value but a segmentation map with the same dimensions as the input image. Think about why convolutional layers are preferable over fully-connected layers for this particular task.

## Loss and Metrics
In the context of image segmentation, the loss function employed is pixel-wise cross entropy loss. This particular loss function operates at the pixel level, considering each pixel individually, comparing the class predictions (depth-wise pixel vector) to our one-hot encoded target vector. 
[Figure omitted from code-only export.]
source: https://www.jeremyjordan.me/semantic-segmentation/

Up until now we only used the default loss function (`nn.CrossEntropyLoss`) in our solvers. However, In order to ignore the `unlabeled` pixels for the computation of our loss, we need to use a customized version of the loss for the initialization of our segmentation solver. The `ignore_index` argument of the loss can be used to filter the `unlabeled` pixels and computes the loss only over remaining pixels.


````
loss_func = torch.nn.CrossEntropyLoss(ignore_index=-1, reduction='mean')

for (inputs, targets) in train_data[0:4]:
    inputs, targets = inputs, targets
    outputs = dummy_model(inputs.unsqueeze(0))
    losses = loss_func(outputs, targets.unsqueeze(0)).item()
    print("Loss:", losses)
````

<div class="alert alert-warning">

<h3>Note: Non-zero loss for the first sample</h3>

The output of our dummy model is one-hot-coded tensor. Since <b>nn.CrossEntropyLoss</b> applies <b>log_softmax</b> internally, the loss is:

$$
\mathrm{loss}(x, class)
=
-\log \left(
\frac{\exp(x[class])}{\sum_j \exp(x[j])}
\right)
=
-x[class] + \log \left( \sum_j \exp(x[j]) \right)
$$

Therefore, the loss will not be zero.

</div>

To obtain an evaluation accuracy, we can simply compute the average per pixel accuracy of our network for a given image. We will use the following function:

````
def evaluate_model(model, dataloader):
    test_scores = []
    model.eval()
    model = model.to(device)
    for inputs, targets in dataloader:
        inputs, targets = inputs.to(device), targets.to(device)

        outputs = model.forward(inputs)
        _, preds = torch.max(outputs, 1)
        targets_mask = (targets >= 0).cpu()
        test_scores.append(np.mean((preds.cpu() == targets.cpu())[targets_mask].numpy()))

    return np.mean(test_scores)

test_loader = torch.utils.data.DataLoader(test_data, batch_size=1,shuffle=False,num_workers=0)
print(evaluate_model(dummy_model, test_loader))
````

You will see reasonably high numbers as your accuracy when you do the training later. The reason behind that is the fact that most output pixels are of a single class and the network can just overfit to common classes such as "grass".

## Step 2: Train your own model

<div class="alert alert-info">
    <h3>Task: Implement</h3>
    <p> In addition to the network itself, you will also need to write the code for the model training.
    </p>

````
import torch
from exercise_code.networks.segmentation_nn import SegmentationNN
from tqdm import tqdm

def create_tqdm_bar(iterable, desc):
    return tqdm(enumerate(iterable),total=len(iterable), ncols=150, desc=desc)

model = SegmentationNN(hp=hparams)

########################################################################
# TODO - Train Your Model                                              #
########################################################################

from IPython.display import clear_output
import matplotlib.pyplot as plt
from torchvision import transforms

# My comment: data augmentation lives here (not in segmentation_dataset.py,
# per the exercise warning) and is only applied to the training split. Flip
# is applied identically to image and mask; color jitter only touches the
# image since it must not change pixel labels.
class AugmentedSegmentationData(torch.utils.data.Dataset):
    def __init__(self, base_dataset, hflip_p=0.5):
        self.base = base_dataset
        self.hflip_p = hflip_p
        self.color_jitter = transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.2)

    def __len__(self):
        return len(self.base)

    def __getitem__(self, idx):
        img, target = self.base[idx]
        if torch.rand(1).item() < self.hflip_p:
            img = torch.flip(img, dims=[2])
            target = torch.flip(target, dims=[1])
        img = self.color_jitter(img)
        return img, target

# My comment: Build dataloaders here instead of changing the dataset class.
train_loader = torch.utils.data.DataLoader(
    AugmentedSegmentationData(train_data),
    batch_size=8,
    shuffle=True,
    num_workers=0,
    pin_memory=torch.cuda.is_available()
)

val_loader = torch.utils.data.DataLoader(
    val_data,
    batch_size=1,
    shuffle=False,
    num_workers=0
)

# My comment: ignore_index=-1 ignores void pixels in the segmentation masks.
loss_func = torch.nn.CrossEntropyLoss(ignore_index=-1, reduction="mean")

model = model.to(device)

# My comment: the backbone starts from pretrained ImageNet weights and only
# needs light fine-tuning, while the decoder/classifier are randomly
# initialized and need to learn faster - hence the lower backbone LR.
backbone_modules = [model.stem, model.stage1, model.stage2, model.stage3, model.stage4]
backbone_params = [p for module in backbone_modules for p in module.parameters()]
backbone_param_ids = {id(p) for p in backbone_params}
decoder_params = [p for p in model.parameters() if id(p) not in backbone_param_ids]

optimizer = torch.optim.AdamW(
    [
        {"params": backbone_params, "lr": 1e-4},
        {"params": decoder_params, "lr": 1e-3},
    ],
    weight_decay=1e-4
)

scheduler = torch.optim.lr_scheduler.StepLR(
    optimizer,
    step_size=20,
    gamma=0.5
)

num_epochs = 60
best_val_acc = 0.0
best_model_state = None

train_losses = []
train_accs = []
val_accs = []

for epoch in range(num_epochs):
    model.train()
    train_loss = 0.0
    train_acc = 0.0

    train_bar = create_tqdm_bar(train_loader, f"Train, Epoch [{epoch + 1}/{num_epochs}]")

    for batch_idx, (inputs, targets) in train_bar:
        inputs = inputs.to(device)
        targets = targets.to(device).long()

        optimizer.zero_grad()

        outputs = model(inputs)
        loss = loss_func(outputs, targets)

        loss.backward()
        optimizer.step()

        # My comment: Compute pixel accuracy only on labeled pixels.
        with torch.no_grad():
            preds = torch.argmax(outputs, dim=1)
            mask = targets >= 0
            batch_acc = (preds[mask] == targets[mask]).float().mean().item()

        train_loss += loss.item()
        train_acc += batch_acc

        train_bar.set_postfix(
            train_loss=train_loss / (batch_idx + 1),
            train_acc=train_acc / (batch_idx + 1)
        )

    scheduler.step()

    epoch_train_loss = train_loss / len(train_loader)
    epoch_train_acc = train_acc / len(train_loader)
    val_acc = evaluate_model(model, val_loader)

    train_losses.append(epoch_train_loss)
    train_accs.append(epoch_train_acc)
    val_accs.append(val_acc)

    # My comment: Keep the best validation model during training.
    if val_acc > best_val_acc:
        best_val_acc = val_acc
        best_model_state = {
            key: value.cpu().clone()
            for key, value in model.state_dict().items()
        }

    clear_output(wait=True)

    print(
        "Epoch [%d/%d] | train_loss: %.4f | train_acc: %.4f | val_acc: %.4f | best_val_acc: %.4f"
        % (
            epoch + 1,
            num_epochs,
            epoch_train_loss,
            epoch_train_acc,
            val_acc,
            best_val_acc
        )
    )

    plt.figure(figsize=(14, 5))

    plt.subplot(1, 2, 1)
    plt.plot(train_losses, marker="o", label="train loss")
    plt.title("Train Loss: %.4f" % epoch_train_loss)
    plt.xlabel("epoch")
    plt.ylabel("loss")
    plt.grid(True)
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(train_accs, marker="o", label="train acc")
    plt.plot(val_accs, marker="o", label="val acc")
    plt.title(
        "Train Acc: %.4f | Val Acc: %.4f | Best Val: %.4f"
        % (epoch_train_acc, val_acc, best_val_acc)
    )
    plt.xlabel("epoch")
    plt.ylabel("accuracy")
    plt.ylim(0, 1)
    plt.grid(True)
    plt.legend()

    plt.show()

# My comment: Restore the best validation checkpoint before final evaluation.
if best_model_state is not None:
    model.load_state_dict(best_model_state)

model = model.to(device)
print("Best validation accuracy:", best_val_acc)

########################################################################
#                           END OF YOUR CODE                           #
########################################################################
````
